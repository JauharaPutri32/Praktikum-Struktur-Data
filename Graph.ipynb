```python
# Representasi Graf dengan Adjacency List #
graph = {                   #Inisialisasi graf tak berarah (Undirected Graph)
    'A': ['B', 'C'],        #dengan representasi Adjacency List (sebagai dictionary)
    'B': ['A', 'D', 'E'],   #Nama simpul ('A', 'B') bertindak sebagai Key
    'C': ['A', 'F'],        #Daftar simpul yang terhubung (tetangga) disimpan dalam list[] sebagai value
    'D': ['B'],
    'E': ['B', 'F'],
    'F': ['C', 'E']
}

#Menampilkan graph
def print_graph(graph):
    for node in graph:                     #looping pada dict, variabel node akan mengambil setiap Key secara bergantian
        print(f'{node} -> {graph[node]}')  #mencetak Key dan Value

print_graph(graph)  #memanggil fungsi print_graph dengan memasukkan variabel graph sebagai data
```

```python
# Traversal Graph menggunakan BFS #
from collections import deque #Operasi menghapus elemen dari depan (penggunaan deque)

def bfs(graph, start):      #start : simpul awal penelusuran
    visited = set()         #Menyiman simpul yang selesai dikunjungi
    queue = deque([start])  #Prinsip FIFO (masuk pertama, keluar pertama)

    while queue:                      #Looping selama queue tidak kosong
        vertex = queue.popleft()      #Mengambil dan menghapus satu simpul barisan paling depan, lalu disimpan dalam variabel vertex
        if vertex not in visited:     #Apabila simpul vertex belum pernah dikunjungi dalan set visited
            print(vertex, end=' ')    #Mencetak simpul
            visited.add(vertex)       #Memasukkan simpul ke dalam set visited, simpul sudah dikunjungi
            queue.extend([neighbor for neighbor in graph [vertex] if neighbor not in visited])
            #Mencari semua tetangga simpul saat ini yang belum dikunjungi, ditaruh semua sekaligus ke barisan belakang

#Contoh pemanggilan
bfs(graph, 'A')
```

```python
# Traversal Graph menggunakan DFS #
def dfs(graph, start, visited=None):
    if visited is None:
        visited = set()

    if start not in visited:               #Apabila simpul start belum pernah dikunjungi/belum tersimpan dalam set visited
        print(start, end=' ')              #Mencetak simpul
        visited.add(start)                 #Memasukkan simpul ke dalam set visited, simpul sudah dikunjungi
        for neighbor in graph[start]:      #Memeriksa setiap tetangga yang terhubung langsung dengan simpul start
            dfs(graph, neighbor, visited)  #Memanggil fungsi dfs untuk menyelami satu tetangga secara mendalam, sebelum menuju tetangga lain

#Contoh pemanggilan
dfs(graph, 'A')
```

```python
# Latihan Personal #
#Lengkapi fungsi untuk menghitung jumlah tetangga dari tiap simpul dalam graph
def count_neighbors(graph):
    result = {}                           #Membuat dict kosong, tempat menyimpan hasil akhir
    for node in graph:                    #Looping pada dict, variabel node akan mengambil setiap Key secara bergantian
        result[node] = len(graph[node])   #Menghitung elemen di dalam list tetangga, disimpan dalam variabel result sesuai node
    return result                         #Mengembalukan isi dari dict result

#Contoh pemanggilan
neighbor_count = count_neighbors(graph)
print(neighbor_count)
```

```python
## LATIHAN / TUGAS ##
# (1) Graph Tak Berarah #
graph = {
    '1': ['2', '3', '6'],
    '2': ['1', '3', '5'],
    '3': ['1','2', '6'],
    '4': ['5', '6'],
    '5': ['2', '4'],
    '6': ['1','3','4']
}

#Menampilkan graph
def print_graph(graph):
    for node in graph:
        print(f'{node} -> {graph[node]}')

print_graph(graph)
```

```python
## LATIHAN / TUGAS ##
# (2) Implementasi BFS DAN DFS #

#Implementasi BFS
from collections import deque

def bfs(graph, start):
    visited = set()
    queue = deque([start])

    while queue:
        vertex = queue.popleft()
        if vertex not in visited:
            print(vertex, end=' ')
            visited.add(vertex)
            queue.extend([neighbor for neighbor in graph [vertex] if neighbor not in visited])

#Contoh pemanggilan
print('Hasil BFS: ')
bfs(graph, '1')
print('\n-----------')

#Implementasi DFS
def dfs(graph, start, visited=None):
    if visited is None:
        visited = set()

    if start not in visited:
        print(start, end=' ')
        visited.add(start)
        for neighbor in graph[start]:
            dfs(graph, neighbor, visited)

#Contoh pemanggilan
print('Hasil DFS: ')
dfs(graph, '1')
```

```python
## LATIHAN / TUGAS ##
# (3) Modifikasi fungsi BFS agar mengembalikan list urutan kunjungan #

def bfs_list(graph, start):
    visited = set([start])   #Catatan simpul yang sudah/akan dikunjungi
    queue = deque([start])   #Antrean untuk urutan pemeriksaan (FIFO)
    urutan_kunjungan = []    #Wadah hasil akhir urutan BFS

    while queue:                          #Looping selama antrean masih ada isi
        simpul = queue.popleft()          #Ambil simpul terdepan
        urutan_kunjungan.append(simpul)   #Catat ke hasil akhir

        for tetangga in graph[simpul]:    #Cek semua tetangga simpul aktif
            if tetangga not in visited:   #Jika tetangga belum terdaftar
                visited.add(tetangga)     #Tandai sebagai dikunjungi
                queue.append(tetangga)    #Masukkan ke antrean belakang
    return urutan_kunjungan               #Kembalikan list hasil akhir

hasil_bfs = bfs_list(graph, '1')
print(hasil_bfs)
```

```python
## LATIHAN / TUGAS ##
# (4) Fungsi yang mengembalikan satu jalur dari start ke end menggunakan DFS #

def find_path(graph, start, end, visited=None, path=None):
    if visited is None:
        visited = set()     #Tempat mencatat simpul yang sudah dilewati
    if path is None:
        path = []           #List untuk merekam rute perjalanan saat ini

    visited.add(start)      #Tandai simpul sekarang agar tidak berputar-putar
    path = path + [start]   #Masukkan simpul sekarang ke dalam rute jalan

    if start == end:
        return path         #KONDISI BERHENTI: Jika sudah sampai tujuan, kirim jalurnya

    for tetangga in graph[start]:   #Cek jalan ke setiap tetangga simpul aktif
        if tetangga not in visited: #Jika jalan ke tetangga belum pernah dicoba
        #Selami tetangga tersebut secara mendalam (rekursif)
            new_path = find_path(graph, tetangga, end, visited, path)
            if new_path:
                return new_path     #Jika jalur sukses tembus ke tujuan, langsung oper ke atas

    return None                     #Jika semua tetangga mentok dan tidak tembus ke tujuan

jalur = find_path(graph, '1', '5') #Cari satu jalur dari simpul '1' ke '5'
print(f"Jalur ditemukan: {' -> '.join(jalur) if jalur else 'Tidak ada jalur'}")
```

```python
## LATIHAN / TUGAS ##
# (5) Fungsi untuk mengecek simpul dalam graph saling terhubung #

def is_connected(graph):
    if not graph:
        return True         #Jika graf kosong, otomatis dianggap terhubung

    #Ambil simpul pertama dari graf sebagai titik awal penjelajahan
    start_node = list(graph.keys())[0]
    visited = set()         #Tempat mencatat semua simpul yang berhasil dijangkau

    # Fungsi bantuan untuk menjelajah graf secara mendalam (DFS)
    def dfs_helper(simpul):
        visited.add(simpul)             #Tandai simpul saat ini sebagai terjangkau
        for tetangga in graph[simpul]:
            if tetangga not in visited:
                dfs_helper(tetangga)    #Selami tetangga yang belum dikunjungi

    dfs_helper(start_node)  #Jalankan DFS dari simpul pertama tadi

    #KUNCI LOGIKA: Terhubung jika jumlah simpul terjangkau SAMA DENGAN total semua simpul
    return len(visited) == len(graph)

is_connected = is_connected(graph)
print(f"Graph {'terhubung' if is_connected else 'tidak terhubung'}")
```

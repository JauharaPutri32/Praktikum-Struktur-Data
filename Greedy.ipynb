```python
# Case 1 : Coin Change #
def coin_change_greedy(amount, coins):  #amount : jumlah uang yang ingin dikembalikan, coins : daftar denominasi koin
    coins.sort(reverse=True)            #Mengurutkan daftar koin dari nilai terbesar ke terkecil
    result = []                         #Menympan koin yang berhasil dipilih untuk kembalian

    for coin in coins:            #Memeriksa setiap koin dalam daftar satu per satu
        while amount >= coin:     #Apabila nilai amount lebih besar sama dengan nilai koin saat ini
            amount -= coin        #Mengurangi nilai amount dengan nilai koin saat ini
            result.append(coin)   #Memasukkan koin yang terpilih ke dalam daftar
    return result                 #Mengembalikan daftar koin yang berhasil dipilih

#Contoh penggunaan
amount = 57
coins = [25, 10, 5, 1]
change = coin_change_greedy(amount, coins)
print("Koin yang digunakan:", change)
```

```python
# Fractional Knapsack #
def fractional_knapsack(items, capacity):  #items = list of tuples (nilai, berat)
  items = sorted(items, key=lambda x: x[0]/x[1], reverse=True)
  #Mengurutkan daftar barang secara menurun berdasarkan rasio nilai per berat
  total_value = 0.0                        #Menampung total nilai barang yang berhasil masuk ke ransel
  #tipe data float

  for value, weight in items:                      #Mengecek setiap barang
    if capacity >= weight:                         #Apabila nilai capacity masih cukup menampung berat barang saat ini
      capacity -= weight                           #Mengurangi nilai capacity dengan berat barang saat ini
      total_value += value                         #Menambahkan nilai barang saat ini ke total nilai
    else:
      total_value += value * (capacity / weight)   #Mengambil barang yang muat dengan sisa kapasitas yang ada
      break                                        #Perulangan berhenti, wadah dipastikan sudah terisi penuh
  return total_value

#Contoh penggunaan
items = [(60, 10), (100, 20), (120, 30)]
capacity = 50
max_value = fractional_knapsack(items, capacity)
print("Nilai maksimum yang dapat dibawa:", max_value)
```

```python
# Minimum Spanning Tree: Prim dan Kruskal
#Contoh kode program 1:
import heapq   #Membuat struktur data antrean prioritas, elemen dengan bobot terkecil selalu di urutan paling atas untuk diambil terlebih dahulu

def prim_mst(graph, start):
  visited = set()             #Membuat set kosong untuk mencatat simpul
  min_heap = [(0, start)]     #Membuat list dengan tuple (bobot, simpul)
  total_weight = 0            #Menyimpan akumulasi total bobot dari MST yang terbentuk

  while min_heap:                               #Berjalan selama masih ada kandidat sisi yang dapat diperiksa
    weight, node = heapq.heappop(min_heap)      #Mengambil elemen dengan bobot terkecil, heappop : menjamin selalu dapat sisi paling ringan
    if node in visited:                         #Pengecekan apabila simpul yang diambil sudah ada di dalam set
      continue                                  #Maka abaikan dan lanjut ke perulangan berikutnya, untuk menghindari siklus
    visited.add(node)                           #Masukkan simpul ke set apabila simpul belum pernah dikunjungi
    total_weight += weight                      #Tambahkan bobotnya ke variabel total_weight

    for neighbor, edge_weight in graph[node]:               #Memeriksa semua tetangga dan bobot sisinya yang terhubung langsung dengan simpul yang baru dikunjungi
      if neighbor not in visited:
        heapq.heappush(min_heap, (edge_weight, neighbor))   #Masukkan tetangga beserta bobot sisinya ke min_heap yang otomatis mengatur posisinya berdasarkan bobot terkecil
  return total_weight

#Representasi graph: adjacency list
graph = {
    'A': [('B', 2), ('C', 3)],
    'B': [('A', 2), ('C', 1), ('D', 1)],
    'C': [('A', 3), ('B', 1), ('D', 4)],
    'D': [('B', 1), ('C', 4)]
}
print("Total bobot MST (Prim) :", prim_mst(graph, 'A'))
```

```python
# Minimum Spanning Tree: Prim dan Kruskal
#Contoh kode program 2:
def kruskal_mst(edges, n_nodes):             #n_nodes : jumla total simpul dalam graph
#edges : daftar sisi berupa tuple berisi simpul asal, simpul tujuan, dan bobot)
  parent = {i: i for i in range(n_nodes)}    #Membuat dict, di awal setiap simpul bertindak sebagai parent
  #Contoh untuk 4 simpul (0 - 3) : {0: 0, 1: 1, 2: 2, 3: 3}

  #Mencari akar/root dari simpul x
  def find(x):
    while parent[x] != x:
      x = parent[x]
    return x

  def union(x, y):               #Menggabungkan dua simpul ke dalam satu grup yang sama
    root_x = find(x)             #Mencari akar simpul x
    root_y = find(y)             #Mencari akar simpul y
    if root_x != root_y:         #Apabila akarnya berbeda, simpul x dan y berada di komponen terpisah
      parent[root_y] = root_x    #Gabungkan dengan mengubah parent root_ y menjadi root_x
      return True
    return False                 #Apabila akarnya sama, diabaikan karena apabila dihubungkan akan tercipta siklus

  edges.sort(key=lambda x: x[2])    #Mengurutkan daftar sisi dari bobot terkecil ke terbesar
  #Indeks [2] merujuk pada elemen ketiga dari tuple (u, v, weight)
  total_weight = 0

  for u, v, weight in edges:  #Memeriksa setiap sisi yang sudah diurutkan
    if union(u, v):           #Apabila fungsi union(u, v) mengembalikan nilai True
      total_weight += weight
  return total_weight

edges = [(0, 1, 2), (0, 2, 3), (1, 2, 1), (1, 3, 1), (2, 3, 4)]
print("Total bobot MST (Kruskal) :", kruskal_mst(edges, 4))
```

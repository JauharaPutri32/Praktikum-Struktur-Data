```python
# Hash Table Sederhana #
class SimpleHashTable:
    def __init__(self, size):     #size = ukuran tabel
      self.size = size            #Menyimpan ukuran tabel
      self.table = [None] * size  #Membuat array kosong dengan panjang sesuai size

    #Mengambil sisa bagi dengan ukiran tabel agar hasil sesuai dengan indeks array
    def hash_function(self, key):   #Fungsi hash bawaan python
      return hash(key) % self.size  #hash(key) : mengubah teks/objek menjadi angka acak yang unik

    #Memasukkan value ke indeks yang ditentukan oleh fungsi hash dari key
    def insert(self, key, value):
      index = self.hash_function(key)                     #Mencari indeks menggunakan fungsi hash
      if self.table[index] is not None:                   #Apabila slot tabel tidak kosong/sudah terdapat isinya
        print(f'Tabrakan terjadi pada indeks {index}')
      else:
        print(f'Data {key} masuk ke indeks {index}')
      self.table[index] = value                           #self.table = menyimpan data, ke dalam slot tabel pada indeks tersebut

    #Mengakses data berdasarkan key yang di-hash
    def search(self, key):
      index = self.hash_function(key) #Mencari indeks data dengan key tertentu disimpan
      return self.table[index]        #Mengembalikan data/nilai di slot

#Contoh penggunaan
ht = SimpleHashTable(10)
ht.insert("apel", 100)
ht.insert("kiwi", 100)
ht.insert("pisang", 200)

print('--------------------------------')
print("Nilai untuk 'Apel'\t:", ht.search("apel"))
print("Nilai untuk 'Pisang'\t:", ht.search("pisang"))
```

```python
# Penanganan Collision dengan Linear Probing #
class LinearProbingHashTable:
  def __init__(self, size):
    self.size = size
    self.table = [None] * size

  def hash_function(self, key):
    return hash(key) % self.size

  def insert(self, key, value):
    index = self.hash_function(key)
    original_index = index                     #Mencatat indeks awal

    while self.table[index] is not None:       #Slot pada indeks tertentu sudah ada isinya
      if self.table[index][0] == key:          #Apabila key pada slot sama dengan key
        break                                  #perulangan berhenti, proses update data
      index = (index + 1) % self.size          #Key-nya berbeda, pergeseran ke slot selanjutnya
      #% self.size : memastikan pencarian sampai ke ujung tabel hingga kembali ke indeks 0 (berputar)
      if index == original_index:              #Apabila perulangan sudah kembali ke indeks awal
        raise Exception("Hash Table penuh")    #berarti slot sudah terisi penuh sehingga memunculkan kondisi exception

    self.table[index] = (key, value)           #Setelah mendapat slot, simpan data dalam bentuk pasangan key value

  def search(self, key):
    index = self.hash_function(key)
    original_index = index

    while self.table[index] is not None:
      if self.table[index][0] == key:
        return self.table[index][1]        #Mengembalikan value-nya
      index = (index + 1) % self.size      #Apabila ketika insert terjadi tabrakan
      if index == original_index:
        break
    return None                            #Mengembalikan None

#Contoh penggunaan
ht = LinearProbingHashTable(5)
ht.insert("apel", 100)
ht.insert("pisang", 200)
ht.insert("melon", 300)
ht.insert("kiwi", 400)

print("Nilai untuk 'Apel'\t:", ht.search("apel"))
print("Nilai untuk 'Melon'\t:", ht.search("melon"))

print('--------------------------------------------')
print(f"Index 'Apel'  : {hash('apel')} % 5 = {hash('apel') % 5}")
print(f"Index 'Pisang': {hash('pisang')} % 5 = {hash('pisang') % 5}")
print(f"Index 'Melon' : {hash('melon')} % 5 = {hash('melon') % 5}")
print(f"Index 'Kiwi'  : {hash('kiwi')} % 5 = {hash('kiwi') % 5}")

print("------------ Hash Table/Storage ------------")
for i in range(ht.size):
    print(f"Index [{i}]: {ht.table[i]}")
```

```python
# Penanganan Collision dengan Chaining #

class ChainingHashTable:
  def __init__(self, size):
    self.size = size
    self.table = [[] for _ in range(size)]  #Membuat array berisi list kosong dengan panjang sesuai size

  def hash_function(self, key):
    return hash(key) % self.size

  def insert(self, key, value):
    index = self.hash_function(key)
    for i, (k, v) in enumerate(self.table[index]):  #Memeriksa isi di dalam hash table pada index tertentu
    #enumerate() : mengetahui isi pasangan key value dan indeksnya
      if k == key:                                  #Kunci baru sama dengan key yang ada pada hash table
        self.table[index][i] = (key, value)         #Proses update data, mengganti nilai lama dengan nilai baru
        return
    self.table[index].append((key, value))          #Apabila terjadi tabrakan, program akan memasukkan data baru ke bagian paling belakang dari list

  def search(self, key):
    index = self.hash_function(key)
    for k, v in self.table[index]:    #Mencari seluruh isi data yang berada di hash tabel satu per satu dari depan ke belakang
      if k == key:
        return v                      #Mengembalikan nilai
    return None                       #Kunci tidak ditemukan, mengembalikan None

#Contoh penggunaan
ht = ChainingHashTable(5)
ht.insert("apel", 100)
ht.insert("pisang", 200)
ht.insert("melon", 300)
ht.insert("kiwi", 400)

print("Nilai untuk 'Apel':", ht.search("apel"))
print("Nilai untuk 'Melon':", ht.search("melon"))

print('--------------------------------------------')
print(f"Index 'Apel'  : {hash('apel')} % 5 = {hash('apel') % 5}")
print(f"Index 'Pisang': {hash('pisang')} % 5 = {hash('pisang') % 5}")
print(f"Index 'Melon' : {hash('melon')} % 5 = {hash('melon') % 5}")
print(f"Index 'Kiwi'  : {hash('kiwi')} % 5 = {hash('kiwi') % 5}")

print("------------ Hash Table/Storage ------------")
for i in range(ht.size):
    print(f"Index [{i}]: {ht.table[i]}")
```

```python
## LATIHAN/TUGAS ##
# (1) Eksperimen Sedeerhana Linear Probing #

class LinearProbingHashTable:
  def __init__(self, size):
    self.size = size
    self.table = [None] * size

  def hash_function(self, key):
    return hash(key) % self.size

  def insert(self, key, value):
    index = self.hash_function(key)
    original_index = index

    while self.table[index] is not None:
      if self.table[index][0] == key:
        break
      index = (index + 1) % self.size
      if index == original_index:
        raise Exception("Hash Table penuh")

    self.table[index] = (key, value)

  def search(self, key):
    index = self.hash_function(key)
    original_index = index

    while self.table[index] is not None:
      if self.table[index][0] == key:
        return self.table[index][1]
      index = (index + 1) % self.size
      if index == original_index:
        break
    return None

#Contoh penggunaan
ht = LinearProbingHashTable(5)
ht.insert("A", 10)
ht.insert("B", 20)
ht.insert("C", 30)
ht.insert("D", 40)
ht.insert("E", 50)

print('--------------------------------------------')
print(f"Index 'A' : {hash('A')} % 5 \t= {hash('A') % 5}")
print(f"Index 'B' : {hash('B')} % 5 \t= {hash('B') % 5}")
print(f"Index 'C' : {hash('C')} % 5 \t= {hash('C') % 5}")
print(f"Index 'D' : {hash('D')} % 5 \t= {hash('D') % 5}")
print(f"Index 'E' : {hash('E')} % 5 \t= {hash('E') % 5}")

print("------------ Hash Table/Storage ------------")
for i in range(ht.size):
    print(f"Index [{i}]: {ht.table[i]}")
```

```python
## LATIHAN/TUGAS ##
# (2) Percobaan Dasar Chaining #

class ChainingHashTable:
  def __init__(self, size):
    self.size = size
    self.table = [[] for _ in range(size)]

  def hash_function(self, key):
    return hash(key) % self.size

  def insert(self, key, value):
    index = self.hash_function(key)
    for i, (k, v) in enumerate(self.table[index]):
      if k == key:
        self.table[index][i] = (key, value)
        return
    self.table[index].append((key, value))

  def search(self, key):
    index = self.hash_function(key)
    for k, v in self.table[index]:
      if k == key:
        return v
    return None

#Contoh penggunaan
ht = ChainingHashTable(3)
ht.insert("apel", 100)
ht.insert("melon", 200)
ht.insert("lemon", 300)
ht.insert("pisang", 400)

print('----------------------------------------------------')
print(f"Index 'Apel'   : {hash('apel')} % 3 \t= {hash('apel') % 3}")
print(f"Index 'Melon'  : {hash('melon')} % 3 \t= {hash('melon') % 3}")
print(f"Index 'Lemon'  : {hash('lemon')} % 3 \t= {hash('lemon') % 3}")
print(f"Index 'Pisang' : {hash('pisang')} % 3 \t= {hash('pisang') % 3}")

print("---------------- Hash Table/Storage ----------------")
for i in range(ht.size):
    print(f"Index [{i}]: {ht.table[i]}")
```

```python
# Pencarian Linier #
def LinierSearch(arr, N, x): #Fungsi LinierSearch(array, panjang array, target)

  for i in range(0, N): #Perulangan index dari 0 hingga N-1
    if (arr[i] == x):   #Pemeriksaan setiap elemen pada index tertentu sama dengan target atau tidak
      return i          #Kondisi sesuai: fungsi mengembalikan index pada elemen yang nilainya sama dengan target
  return -1             #Perulangan tidak menemukan data yang sesuai

#Kode utama
if __name__ == "__main__":
  arr = [20, 10, 30, 50, 40] #Berupa array yang elemennya akan dicari
  x = 10                     #Berupa kunci/target yang ingin dicari posisinya
  N = len(arr)               #Menghitung jumlah elemen

  #Pemanggilan fungsi
  result = LinierSearch(arr, N, x)
  if(result == -1): #Saat fungsi tidak menemukan data yang sesuai dan mengembalikan -1
    print("Element yang di cari tidak ada dalam array")
  else:
    print("Element yang di cari berada pada indeks ke-", result)
```

```python
# Binary Search(iteratif) #
def BinarySearch(arr, l, r, x): #Fungsi BinarySearch(array, left/kiri, right/kanan, kunci)

  while l <= r:             #Perulangan berjalan selama sisi kiri masih ada
    mid = l + (r - l) // 2  #Mencari titik tengah (angka bulat)

    #Pemeriksaan jika kunci berada ditengah
    if arr[mid] == x:
      return mid       #Mengembalikkan index elemen yang sesuai
    #Apabila angka tengah lebih kecil dari kunci, kunci berada di kanan
    elif arr[mid] < x:
      l = mid + 1      #Menggeser batas kiri satu posisi setelah titik tengah
    #Apabila angka tengah lebih besar dari kunci, kunci berada di kiri
    else:
      r = mid - 1      #Menggeser batas kanan satu posisi sebelum titik tengah

  return -1                 #Perulangan tidak menemukan data yang sesuai

#Kode utama
if __name__ == "__main__":
  arr = [2, 5, 8, 12, 16, 23, 38, 56, 72, 91]  #Array harus dalam kondisi urut
  x = 72

  #Pemanggilan fungsi
  result = BinarySearch(arr, 0, len(arr)-1, x)   #Pemeriksaan dimulai dari index 0 hingga N-1
  if result != -1:
    print("Element terdapat pada indeks ke-", result)
  else:
    print("Element tidak ada dalam array")
```

```python
# Binary Search(rekursif) #
def BinarySearch(arr, l, r, x):

  if r >= l:
    mid = l + (r - l) // 2

    #Apabila kunci di tengah
    if arr[mid] == x:
      return mid
    #Apabila angka tengah lebih kecil dari kunci, kunci berada di kanan
    elif arr[mid] < x:
      return BinarySearch(arr, mid + 1, r, x)  #Memanggil fungsi dan mengubah batas kiri menjadi mid+1
    #Apabila angka tengah lebih besar dari kunci, kunci berada di kiri
    else:
      return BinarySearch(arr, l, mid - 1, x)  #Memanggil fungsi dan mengubah batas kanan menjadi mid-1

  #Apabila r lebih kecil daripada l, data tidak ditemukan
  else:
    return -1

#Kode utama
if __name__ == "__main__":
  arr = [2, 5, 8, 12, 16, 23, 38, 56, 72, 91]
  x = 72

  #Pemanggilan fungsi
  result = BinarySearch(arr, 0, len(arr)-1, x)
  if result != -1:
    print("Element terdapat pada indeks ke-", result)
  else:
    print("Element tidak ada dalam array")
```

```python
# Studi Kasus 1 #

class pair:
  def __init__(self):
    self.min = 0       #Menyimpan angka terkecil
    self.max = 0       #Menyimpan angka terbesar

def getMinMax(arr, N):
  minmax = pair()      #Menampung hasil akhir

  #Apabila isi array hanya satu, maka elemen tersebut menjadi min and max
  if N == 1:
    minmax.max = arr[0]
    minmax.min = arr[0]
    return minmax  #Mengembalikan elemen tersebut

  #Apabila isi array lebih dari satu, gunakan perbandingan
  if arr[0] > arr[1]:         #Jika angka pertama lebih besar dibanding angka kedua
    minmax.max = arr[0]       #Angka pertama menjadi max
    minmax.min = arr[1]       #Angka kedua menjadi min
  else:
    minmax.max = arr[1]
    minmax.min = arr[0]
  #mengecek angka ketiga hingga ujung array
  for i in range(2, N):
    if arr[i] > minmax.max:   #Jika angka pada index tertentu lebih besar dari angka terbesar saat ini
      minmax.max = arr[i]     #maka angka terbesar saat ini digantikan dengan yang lebih besar
    elif arr[i] < minmax.min: #Jika angka pada index tertentu lebih kecil dari angka terkecil saat ini
      minmax.min = arr[i]     #maka angka terkecil saat ini digantikan dengan yang lebih kecil
  return minmax

#Kode utama
if __name__ == "__main__":
  arr = [1000, 11, 445, 1, 330, 3000]
  N = 6
  minmax = getMinMax(arr, N)
  print("Minimum element is", minmax.min)
  print("Maximum element is", minmax.max)
```

```python
# Studi Kasus 2 #

def printTwoElements(arr):
  n = len(arr)
  temp = [0] * n        #Membuat array berisi nilai 0 sebanyak panjang array (n)
  repeatingNumber = -1  #Menyimpan angka yang bernilai ganda
  missingNumber = -1    #Menyimpan angka yang hilang

  #Pemeriksaan satu persatu
  for i in range(n):
    temp[arr[i] - 1] += 1
    if temp[arr[i] - 1] > 1:    #Apabila array temp trdpt angka lebih dari satu kali muncul
      repeatingNumber = arr[i]  #angka akan tersimpan

  for i in range(n):
    if temp[i] == 0:            #Apabila array dengan semua isi 0 masih terdapat index yang elemennya 0
      missingNumber = i + 1     #index ditambah 1, angka tersimpan
      break

  print("The repeating number is", repeatingNumber, ".")
  print("The missing number is", missingNumber, ".")

arr = [7, 3, 4, 5, 5, 6, 2]
printTwoElements(arr)
```

```python
# Studi Kasus 3 #

def findCouple(arr, x):
    find = set()  #Menyimpan angka yang terlihat

    for a in arr:
        target = x - a  #Untuk menemukan pasangan jumlah a yang menghasilkan x
        if target in find:
            print(f"Terdapat pasangan {a, target} dengan jumlah {x}")
            return True

        find.add(a)     #Untuk pengecekan
    return False        #Tidak menemukan pasangan

#Kode utama
arr = [11, 15, 26, 38, 9, 10]
x = 35

if not findCouple(arr, x):
    print("Tidak ada pasangan yang cocok")
```

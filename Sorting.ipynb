```python
# Bubble Sort #

def bubble_sort(arr):                       #Fungsi bubble_sort(array)
  n = len(arr)                              #Panjang array
  for i in range(n-1):                      #Apabila jumlah putaran/perulangan sudah memenuhi n-1
    for j in range(n-i-1):                  #Menentukan jumlah perbandingan yang dilakukan
      if arr[j] > arr[j+1]:                 #Dua elemen dibandingkan, dilakukan sebanyak jumlah perbandingan
       arr[j], arr[j+1] = arr[j+1], arr[j]  #Menukar posisi, angka yang lebih besar di sisi kanan
  return arr                                #mengembalikan array dalam kondisi urut

#Contoh penggunaan
data = [64, 34, 25, 12, 22, 11, 90]
sorted_data = bubble_sort(data.copy()) #Menggunakan copy agar data asli tidak berubah
print("Data sebelum sorting:", data)
print("Data setelah Bubble Sort:", sorted_data)
```

```python
# Selection Sort #

def selection_sort(arr):                             #Fungsi selection_sort(array)
  n = len(arr)
  for i in range(n-1):
    min_index = i                                    #Pemisalan nilai i adalah angka yang paling kecil, tersimpan pada min_index
    for j in range(i+1, n):                          #Pencarian angka terkecil dimulai dari sisi kanan nilai i hingga akhir array
      if arr[j] < arr[min_index]:                    #Apabila nilai j lebih kecil dibanding nilai min_index
        min_index = j                                #Maka nilai min_index diganti dengan nilai j
    arr[i], arr[min_index] = arr[min_index], arr[i]  #Menukar posisi, angka yang lebih kecil di sisi kiri
  return arr

#Contoh penggunaan
data = [64, 25, 12, 22, 11]
sorted_data = selection_sort(data.copy())
print("Data sebelum sorting:", data)
print("Data setelah Selection Sort:", sorted_data)
```

```python
# Insertion Sort #

def insertion_sort(arr):             #Fungsi insertion_sort(array)
  n = len(arr)
  for i in range(1, n):              #Perulangan dimulai dari index 1 hingga akhir array
    key = arr[i]                     #Angka pada index tertentu disimpan pada variabel key
    j = i - 1                        #Variabel perbandingan dimulai dari sisi kiri nilai i
    while j >= 0 and arr[j] > key:   #Perulangan berjalan ke sisi kiri, di mana nilai j lebih besar dari key
      arr[j+1] = arr[j]              #Geser ke sisi kanan apabila nilai pada sisi kiri lebih besar
      j -= 1                         #Berlanjut ke arah kiri untuk membandingkan key dengan angka berikutnya
    arr[j+1] = key                   #Sisi kiri tidak lagi lebih besar dari key, letakkan key pada posisi tersebut
  return arr

#Contoh penggunaan
data = [64, 34, 25, 12, 22, 11, 90]
sorted_data = insertion_sort(data.copy())
print("Data sebelum sorting:", data)
print("Data setelah Insertion Sort:", sorted_data)
```

```python
# Merge Sort #

def merge_sort(arr):        #Fungsi merge_sort(array)
  if len(arr) > 1:          #Berjalan selama panjang array lebih dari satu
    mid = len(arr) // 2     #Mencari titik tengah panjang array dalam bentuk bulat
    left_half = arr[:mid]   #Mengambil sisi kiri array
    right_half = arr[mid:]  #Mengambil sisi kanan array
    merge_sort(left_half)   #Rekusi memanggil fungsi diri sendiri bagian kiri
    merge_sort(right_half)  #dan bagian kanan hingga elemennya berdiri sendiri

    i = j = k = 0                                      #i untuk left_half, j untuk right_half, k untuk array
    while i < len(left_half) and j < len(right_half):  #Perbandingan berulang dari kedua sisi
      if left_half[i] < right_half[j]:                 #Apabila sisi kiri lebih kecil dari sisi kanan
        arr[k] = left_half[i]                          #Angka yang lebih kecil ditambahkan ke array
        i += 1                                         #Berlanjut seiring bertambahnya index i
      else:
        arr[k] = right_half[j]                         #Selain kondisi di atas, angka yang lebih besar ditambahkan ke array
        j += 1                                         #Berlanjut seiring bertambahnya index j
      k += 1                                           #Berlanjut ke perbandingan selanjutnya, berhenti apabila salah satu sisi kosong

    while i < len(left_half):  #Perulangan apabila sisi kiri terdapat sisa (array ganjil)
      arr[k] = left_half[i]    #Di posisikan pada array left_half secara berurutan
      i += 1
      k += 1

    while j < len(right_half): #Perulangan apabila sisi kanan terdapat sisa (array ganjil)
      arr[k] = right_half[j]   #Di posisikan pada array right_half secara berurutan
      j += 1
      k += 1
  return arr

#Contoh penggunaan
data = [64, 25, 12, 22, 11]
sorted_data = merge_sort(data.copy())
print("Data sebelum sorting:", data)
print("Data setelah Merge Sort:", sorted_data)
```

```python
# Quick Sort #

def quick_sort(arr):  #Fungsi quick_sort(array)
  if len(arr) <= 1:   #Apabila panjang array hanya satu/kosong
    return arr        #fungsi akan mengembalikan array

  pivot = arr[len(arr) // 2]                           #Mencari nilai tengah array
  left = [x for x in arr if x < pivot]                 #Menyimpan sisi yang nilainya lebih kecil daripada nilai pivot
  middle = [x for x in arr if x == pivot]              #Menyimpan sisi yang nilainya sama dengan nilai pivot
  right = [x for x in arr if x > pivot]                #Menyimpan sisi yang nilainya lebih besar daripada nilai pivot
  return quick_sort(left) + middle + quick_sort(right) #Penggabungan ketiga sisi

#Contoh penggunaan
data = [64, 25, 12, 22, 11]
sorted_data = quick_sort(data.copy())
print("Data sebelum sorting:", data)
print("Data setelah Quick Sort:", sorted_data)
```

```python
## LATIHAN / TUGAS ##
# Optimasi Pencarian Produk di Marketplace #

def quick_sort(list):
  if len(list) <= 1:
    return list

  pivot = list[len(list) // 2]
  left = [x for x in list if x [1] < pivot[1]]
  middle = [x for x in list if x [1] == pivot[1]]
  right = [x for x in list if x [1] > pivot[1]]
  return quick_sort(left) + middle + quick_sort(right)

#Berdasarkan harga terendah hingga tertinggi
daftar_snack = {'Fullo' : 500, 'Keju Cake' : 1500, 'Taro' : 2000, 'Pilus' : 1000, 'Tik Tak' : 1500, 'Regal': 1500}
items = list(daftar_snack.items())
sorted_data = quick_sort(items)
print("Daftar snack :", sorted_data)

#Berdasarkan rating pelanggan
daftar_minuman = {'Floridina' : 5, 'Ultramilk' : 3, 'Milo' : 2, 'Bua Vita' : 5, 'Teh kotak' : 5, 'Pocari' : 3}
items = list(daftar_minuman.items())
sorted_data = quick_sort(items)
print("Daftar minuman :", sorted_data)
```

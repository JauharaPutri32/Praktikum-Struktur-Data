```python
# Binary Tree #
class node:        #definisi struktur node
  def __init__(self, val):
    self.v = val   #menyimpan value/data
    self.l = None  #penunjuk/pointer ke anak sebelah kiri
    self.r = None  #penunjuk/pointer ke anak sebelah kanan

#Membuat tree secara manual
root = node(1)      #membuat node utama bernilai 1
root.l = node(2)    #anak kiri dari node utama bernilai 2
root.r = node(3)    #anak kanan dari node utama bernilai 3
root.l.l = node(4)  #anak kiri dari root.l (2) bernilai 4
root.l.r = node(5)  #anak kanan dari root.l (2) bernilai 5

#Fungsi inorder traversal (left -> root -> right)
def inorder(node):
  if node:
    inorder(node.l)         #menelusuri anak kiri
    print(node.v, end=' ')  #mencetak value
    inorder(node.r)         #menelusuri anak kanan

print("Hasil inorder traversal dari binary tree:")
inorder(root) #memanggil fungsi inorder
```

```python
# Binary Search Tree #
class node:
  def __init__(self, val):
    self.v = val
    self.l = None
    self.r = None

class BinarySearchTree:  #definisi class BST
  def __init__(self):
    self.root = None     #menyimpan tree

  def insert(self, root, val):
    if root is None:    #apabila posisi kosong
      return node(val)  #membuat nilai baru
    if val < root.v:                     #apabila nilai baru lebih kecil dari nilai saat ini
      root.l = self.insert(root.l, val)  #maka akan dimasukkan pada anak kiri
    else:
      root.r = self.insert(root.r, val)  #dimasukkan pada anak kanan
    return root                          #mengembalikan struktur tree

bst = BinarySearchTree()
root = None    #menyimpan hasil pembaruan struktur tree dari setiap iterasi
val_list = [50, 30, 70, 20, 40, 60, 80]
for value in val_list:            #mengambil setiap nilai dari daftar satu persatu
  root = bst.insert(root, value)  #memasukkan nilai ke dalam tree dan memperbarui variabel root
```

```python
# Traversal pada BST #
def inorder(node):
  if node:
    inorder(node.l)
    print(node.v, end=' ')
    inorder(node.r)

def preorder(node):
  if node:
    print(node.v, end=' ')  #mencetak nilai node utama
    preorder(node.l)        #menelusuri sisi kiri
    preorder(node.r)        #menelusuri sisi kanan

def postorder(node):
  if node:
    postorder(node.l)        #menelusuri sisi kiri
    postorder(node.r)        #menelusuri sisi kanan
    print(node.v, end=' ')   #mencetak nilai

print("Inorder Traversal:")
inorder(root)
print("\n\nPreorder Traversal:")
preorder(root)
print("\n\nPostorder Traversal:")
postorder(root)
```

```python
# Pencarian Nilai dalam BST #
def search(node, key):
  if node is None or node.v == key:
    return node
  if key < node.v:
    return search(node.l, key)
  return search(node.r, key)

#Uji pencarian
key = 60
result = search(root, key)
if result:
  print(f"{key} ditemukan dalam tree.")
else:
  print(f"{key} tidak ditemukan.")

key = 25
result = search(root, key)
if result:
  print(f"{key} ditemukan dalam tree.")
else:
  print(f"{key} tidak ditemukan.")
```

```python
## LATIHAN / TUGAS ##
# (2) Program Binary Search Tree #

# Binary Search Tree #
class node:
  def __init__(self, val):
    self.v = val
    self.l = None
    self.r = None

class BinarySearchTree:
  def __init__(self):
    self.root = None

  def insert(self, root, val):
    if root is None:
      return node(val)
    if val < root.v:
      root.l = self.insert(root.l, val)
    else:
      root.r = self.insert(root.r, val)
    return root

bst = BinarySearchTree()
root = None
val_list = [32, 31, 5, 74, 65, 13, 8]
for value in val_list:
  root = bst.insert(root, value)

#--------------------------------------------------#
# Traversal pada BST #
def inorder(node):
  if node:
    inorder(node.l)
    print(node.v, end=' ')
    inorder(node.r)

def preorder(node):
  if node:
    print(node.v, end=' ')
    preorder(node.l)
    preorder(node.r)

def postorder(node):
  if node:
    postorder(node.l)
    postorder(node.r)
    print(node.v, end=' ')

print('------------------------------')
print("-Hasil Traversal-")
print("Inorder Traversal:")
inorder(root)
print("\n\nPreorder Traversal:")
preorder(root)
print("\n\nPostorder Traversal:")
postorder(root)
print('\n------------------------------')

#--------------------------------------------------#
# Pencarian Nilai dalam BST #
def search(node, key):
  if node is None or node.v == key:
    return node
  if key < node.v:
    return search(node.l, key)
  return search(node.r, key)

print("-Hasil Pencarian-")
key = 32
result = search(root, key)
if result:
  print(f"Value {key} ditemukan dalam tree.")
else:
  print(f"Value {key} tidak ditemukan.")

key = 7
result = search(root, key)
if result:
  print(f"\nValue {key} ditemukan dalam tree.")
else:
  print(f"\nValue {key} tidak ditemukan.")
print('------------------------------')
```
